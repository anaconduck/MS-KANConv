# Exploratory Data Analysis (EDA)

Notebook ini berisi visualisasi lengkap untuk publikasi jurnal, mencakup:
1. Distribusi dan Proporsi Kelas Aktivitas
2. Pola Sinyal Sensor Mentah (Time-Series Waveforms) untuk membandingkan karakteristik aktivitas.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
DATA_DIR = '../data/csv'

uci_har_labels = {
    1: 'WALKING', 2: 'WALKING_UPSTAIRS', 3: 'WALKING_DOWNSTAIRS', 
    4: 'SITTING', 5: 'STANDING', 6: 'LAYING'
}

pamap2_labels = {
    1: 'lying', 2: 'sitting', 3: 'standing', 4: 'walking', 5: 'running',
    6: 'cycling', 7: 'Nordic walking', 9: 'watching TV', 10: 'computer work',
    11: 'car driving', 12: 'ascending stairs', 13: 'descending stairs',
    16: 'vacuuming', 17: 'ironing', 18: 'folding laundry', 19: 'house cleaning',
    20: 'playing soccer', 24: 'rope jumping'
}

mhealth_labels = {
    1: 'Standing still', 2: 'Sitting and relaxing', 3: 'Lying down', 4: 'Walking',
    5: 'Climbing stairs', 6: 'Waist bends forward', 7: 'Frontal elevation of arms',
    8: 'Knees bending', 9: 'Cycling', 10: 'Jogging', 11: 'Running', 12: 'Jump front & back'
}

wisdm_labels = {
    'Walking': 'Walking', 'Jogging': 'Jogging', 'Upstairs': 'Upstairs',
    'Downstairs': 'Downstairs', 'Sitting': 'Sitting', 'Standing': 'Standing'
}

def plot_signal(df, label_col, label_val, cols_to_plot, title, num_samples=150):
    subset = df[df[label_col] == label_val].head(num_samples)
    plt.figure(figsize=(10, 4))
    for col in cols_to_plot:
        plt.plot(subset.values[:, df.columns.get_loc(col)], label=col)
    plt.title(title)
    plt.xlabel('Time (samples)')
    plt.ylabel('Amplitude')
    plt.legend(loc='upper right')
    plt.tight_layout()
    plt.show()
```

## 1. UCI-HAR Dataset

```python
uci_path = os.path.join(DATA_DIR, 'uci_har_raw.csv')
if os.path.exists(uci_path):
    print('Loading UCI-HAR dataset...')
    df_uci = pd.read_csv(uci_path, usecols=['label', 'body_acc_x', 'body_acc_y', 'body_acc_z'])
    df_uci['activity'] = df_uci['label'].map(uci_har_labels)
    counts = df_uci['activity'].value_counts()
    
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))
    sns.barplot(x=counts.values, y=counts.index, ax=axes[0], palette='viridis', hue=counts.index, legend=False)
    axes[0].set_title('Distribusi Kelas Aktivitas - UCI HAR')
    axes[0].set_xlabel('Jumlah Sampel (Windows)')
    axes[1].pie(counts.values, labels=counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('viridis', len(counts)))
    axes[1].set_title('Proporsi Kelas Aktivitas - UCI HAR')
    plt.tight_layout()
    plt.show()
    print('Total sampel UCI-HAR:', len(df_uci))
    
    plot_signal(df_uci, 'label', 1, ['body_acc_x', 'body_acc_y', 'body_acc_z'], 'UCI-HAR: Body Acceleration saat WALKING (Dinamis)', num_samples=128)
    plot_signal(df_uci, 'label', 4, ['body_acc_x', 'body_acc_y', 'body_acc_z'], 'UCI-HAR: Body Acceleration saat SITTING (Statis)', num_samples=128)
else:
    print('File uci_har_raw.csv tidak ditemukan.')
```

## 2. PAMAP2 Dataset

```python
pamap_path = os.path.join(DATA_DIR, 'pamap2_raw.csv')
if os.path.exists(pamap_path):
    print('Loading PAMAP2 dataset...')
    df_pamap = pd.read_csv(pamap_path, usecols=['activity_id', 'hand_acc16g_x', 'hand_acc16g_y', 'hand_acc16g_z'])
    df_pamap = df_pamap[df_pamap['activity_id'] != 0]
    df_pamap['activity'] = df_pamap['activity_id'].map(pamap2_labels).fillna(df_pamap['activity_id'].astype(str))
    counts = df_pamap['activity'].value_counts()
    
    fig, axes = plt.subplots(1, 2, figsize=(18, 8))
    sns.barplot(x=counts.values, y=counts.index, ax=axes[0], palette='magma', hue=counts.index, legend=False)
    axes[0].set_title('Distribusi Kelas Aktivitas - PAMAP2')
    axes[0].set_xlabel('Jumlah Sampel')
    axes[1].pie(counts.values, labels=counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('magma', len(counts)))
    axes[1].set_title('Proporsi Kelas - PAMAP2')
    plt.tight_layout()
    plt.show()
    print('Total sampel PAMAP2:', len(df_pamap))
    
    plot_signal(df_pamap, 'activity_id', 4, ['hand_acc16g_x', 'hand_acc16g_y', 'hand_acc16g_z'], 'PAMAP2: Hand Acceleration saat WALKING (Dinamis)', num_samples=300)
    plot_signal(df_pamap, 'activity_id', 2, ['hand_acc16g_x', 'hand_acc16g_y', 'hand_acc16g_z'], 'PAMAP2: Hand Acceleration saat SITTING (Statis)', num_samples=300)
else:
    print('File pamap2_raw.csv tidak ditemukan.')
```

## 3. mHealth Dataset

```python
mhealth_path = os.path.join(DATA_DIR, 'mhealth_raw.csv')
if os.path.exists(mhealth_path):
    print('Loading mHealth dataset...')
    df_mhealth = pd.read_csv(mhealth_path, usecols=['activity_id', 'acc_chest_x', 'acc_chest_y', 'acc_chest_z'])
    df_mhealth = df_mhealth[df_mhealth['activity_id'] != 0]
    df_mhealth['activity'] = df_mhealth['activity_id'].map(mhealth_labels).fillna(df_mhealth['activity_id'].astype(str))
    counts = df_mhealth['activity'].value_counts()
    
    fig, axes = plt.subplots(1, 2, figsize=(18, 7))
    sns.barplot(x=counts.values, y=counts.index, ax=axes[0], palette='plasma', hue=counts.index, legend=False)
    axes[0].set_title('Distribusi Kelas Aktivitas - mHealth')
    axes[0].set_xlabel('Jumlah Sampel')
    axes[1].pie(counts.values, labels=counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('plasma', len(counts)))
    axes[1].set_title('Proporsi Kelas - mHealth')
    plt.tight_layout()
    plt.show()
    print('Total sampel mHealth:', len(df_mhealth))
    
    plot_signal(df_mhealth, 'activity_id', 4, ['acc_chest_x', 'acc_chest_y', 'acc_chest_z'], 'mHealth: Chest Acceleration saat WALKING (Dinamis)', num_samples=150)
    plot_signal(df_mhealth, 'activity_id', 2, ['acc_chest_x', 'acc_chest_y', 'acc_chest_z'], 'mHealth: Chest Acceleration saat SITTING (Statis)', num_samples=150)
else:
    print('File mhealth_raw.csv tidak ditemukan.')
```

## 4. WISDM Dataset

```python
wisdm_path = os.path.join(DATA_DIR, 'wisdm_raw.csv')
if os.path.exists(wisdm_path):
    print('Loading WISDM dataset...')
    df_wisdm = pd.read_csv(wisdm_path, usecols=['activity', 'x', 'y', 'z'])
    counts = df_wisdm['activity'].value_counts()
    
    fig, axes = plt.subplots(1, 2, figsize=(18, 7))
    sns.barplot(x=counts.values, y=counts.index, ax=axes[0], palette='coolwarm', hue=counts.index, legend=False)
    axes[0].set_title('Distribusi Kelas Aktivitas - WISDM')
    axes[0].set_xlabel('Jumlah Sampel')
    axes[1].pie(counts.values, labels=counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('coolwarm', len(counts)))
    axes[1].set_title('Proporsi Kelas - WISDM')
    plt.tight_layout()
    plt.show()
    print('Total sampel WISDM:', len(df_wisdm))
    
    # 60 samples = 3 seconds at 20Hz
    plot_signal(df_wisdm, 'activity', 'Walking', ['x', 'y', 'z'], 'WISDM: Acceleration saat WALKING (Dinamis)', num_samples=60)
    plot_signal(df_wisdm, 'activity', 'Sitting', ['x', 'y', 'z'], 'WISDM: Acceleration saat SITTING (Statis)', num_samples=60)
else:
    print('File wisdm_raw.csv tidak ditemukan.')
```
